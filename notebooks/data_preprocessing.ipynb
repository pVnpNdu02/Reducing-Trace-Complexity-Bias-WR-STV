# WR-STV: Raw Data Preprocessing

This notebook parses the raw TrainTicket deployment capture (structured logs,
Prometheus metric JSON dumps, and Jaeger trace JSON dumps) for all three
`ts-order-service` configuration variants into the flat, per-dataset
parquet/CSV files consumed by the main analysis notebook
(`FINAL_REVISED_ANAMOLY_RESEARCH_cleaned_v2.ipynb`).

**Output of this notebook** (written under `<ROOT>/_processed_phase1/<dataset>/`):
- `logs_clean.parquet`, `traces_flat.parquet` -- cleaned raw logs/traces
- `metrics/*.parquet` -- per-container/node resource metric time series
- `spans_long.parquet` -- one row per span; this is the only file the main
  analysis notebook actually reads
- `spans_enriched.parquet` -- span table with tag features and root-to-node
  service-path reconstruction added
- `baseline_stv.npy`, `baseline_stv_trace_ids.csv` -- cached baseline STV matrix

**Raw input required** (not included in this repository -- see `data/README.md`):
the raw TrainTicket deployment capture, one folder per configuration variant,
each containing a structured log CSV, a `potentialAnomalies_*.txt`, a
`Monitoring_*/` directory of Prometheus JSON dumps, and a `Traces_*/`
directory of Jaeger trace JSON dumps.

```python
import os

def show_tree(path, max_items=50):
    path = Path(path)
    print(f"\n=== {path.name} ===")
    items = sorted(list(path.iterdir()))
    for item in items[:max_items]:
        size = item.stat().st_size if item.is_file() else "-"
        print(f"{'[DIR]' if item.is_dir() else '[FILE]':<6} {item.name:<80} {size}")
    if len(items) > max_items:
        print(f"... {len(items)-max_items} more items")

for ds in DATASETS.values():
    show_tree(ROOT / ds)
```

```python
def find_dataset_files(dataset_dir: Path):
    files = {}

    # log structured csv
    structured_logs = list(dataset_dir.glob("LOGS_*_structured.csv"))
    files["structured_log"] = structured_logs[0] if structured_logs else None

    # potential anomalies
    anomaly_txt = list(dataset_dir.glob("potentialAnomalies_*.txt"))
    files["potential_anomalies"] = anomaly_txt[0] if anomaly_txt else None

    # monitoring dir
    monitoring_dirs = [p for p in dataset_dir.iterdir() if p.is_dir() and p.name.startswith("Monitoring_")]
    files["monitoring_dir"] = monitoring_dirs[0] if monitoring_dirs else None

    # traces dir (can be Traces_ or Trace_)
    trace_dirs = [p for p in dataset_dir.iterdir() if p.is_dir() and (p.name.startswith("Traces_") or p.name.startswith("Trace_"))]
    files["traces_dir"] = trace_dirs[0] if trace_dirs else None

    return files

dataset_files = {}

for ds_key, ds_folder in DATASETS.items():
    ds_path = ROOT / ds_folder
    dataset_files[ds_key] = find_dataset_files(ds_path)

    print(f"\n--- {ds_key} ---")
    for k, v in dataset_files[ds_key].items():
        print(f"{k}: {v}")
```

```python
SELECT_METRIC_KEYWORDS = [
    "container_cpu_usage_seconds_total",
    "container_memory_working_set_bytes",
    "container_network_transmit_packets_total",
    "node_cpu_seconds_total",
    "node_memory_MemAvailable_bytes",
]

def pick_metric_files(monitoring_dir: Path, keywords):
    chosen = []
    all_jsons = list(monitoring_dir.glob("*.json"))
    for kw in keywords:
        matches = [p for p in all_jsons if kw in p.name]
        if matches:
            chosen.append(matches[0])
    return chosen

metric_files = {}

for ds_key, files in dataset_files.items():
    chosen = pick_metric_files(files["monitoring_dir"], SELECT_METRIC_KEYWORDS)
    metric_files[ds_key] = chosen

    print(f"\n--- {ds_key} selected metric files ---")
    for p in chosen:
        print(p.name)
```

```python
import json

def load_prometheus_metric_json(json_path):
    with open(json_path, "r", encoding="utf-8", errors="ignore") as f:
        obj = json.load(f)

    rows = []

    # common Prometheus structure:
    # obj["data"]["result"] -> list of metrics
    result = obj.get("data", {}).get("result", [])

    for item in result:
        metric_meta = item.get("metric", {})
        values = item.get("values", [])

        for ts, val in values:
            row = {
                "timestamp": pd.to_datetime(float(ts), unit="s", errors="coerce"),
                "value": pd.to_numeric(val, errors="coerce")
            }
            for mk, mv in metric_meta.items():
                row[f"metric_{mk}"] = mv
            rows.append(row)

    df = pd.DataFrame(rows)
    return df
```

```python
def flatten_trace_json(obj, source_file):
    rows = []

    # Case 1: dict with "data"
    if isinstance(obj, dict) and "data" in obj and isinstance(obj["data"], list):
        items = obj["data"]
    # Case 2: raw list
    elif isinstance(obj, list):
        items = obj
    # Case 3: dict itself is one trace record
    elif isinstance(obj, dict):
        items = [obj]
    else:
        return pd.DataFrame()

    for item in items:
        if isinstance(item, dict):
            flat = pd.json_normalize(item, sep="_")
            flat["source_file"] = source_file
            rows.append(flat)

    if rows:
        return pd.concat(rows, ignore_index=True)
    return pd.DataFrame()
```

```python
import pandas as pd
import numpy as np

def build_process_map_from_row(row):
    """
    Extract mapping like:
    p1 -> ts-order-service
    p2 -> ts-basic-service
    ...
    from columns such as processes_p1_serviceName
    """
    process_map = {}

    for col in row.index:
        if col.startswith("processes_") and col.endswith("_serviceName"):
            pid = col[len("processes_"):-len("_serviceName")]
            process_map[pid] = row[col]

    return process_map
```

```python
def extract_parent_span_id(span):
    """
    Jaeger-like span usually has 'references'.
    We search for CHILD_OF if available.
    """
    refs = span.get("references", [])
    if not refs:
        return None

    for ref in refs:
        if ref.get("refType") == "CHILD_OF":
            return ref.get("spanID")

    # fallback: first reference if no CHILD_OF explicitly found
    if len(refs) > 0:
        return refs[0].get("spanID")

    return None
```

```python
def parse_trace_df_to_spans(trace_df: pd.DataFrame):
    """
    Convert trace-level rows with nested `spans` column
    into long-form span table.
    """
    all_spans = []

    for _, row in trace_df.iterrows():
        trace_id = row.get("traceID")
        source_file = row.get("source_file")

        process_map = build_process_map_from_row(row)

        spans = row.get("spans", [])
        if not isinstance(spans, list):
            continue

        for span in spans:
            if not isinstance(span, dict):
                continue

            span_id = span.get("spanID")
            process_id = span.get("processID")
            parent_span_id = extract_parent_span_id(span)

            record = {
                "traceID": trace_id,
                "spanID": span_id,
                "parentSpanID": parent_span_id,
                "processID": process_id,
                "serviceName": process_map.get(process_id, None),
                "operationName": span.get("operationName"),
                "startTime": span.get("startTime"),
                "duration": span.get("duration"),
                "source_file": source_file,
                "flags": span.get("flags"),
                "warnings_span": span.get("warnings"),
                "tags": span.get("tags"),
                "logs": span.get("logs"),
            }

            all_spans.append(record)

    span_df = pd.DataFrame(all_spans)
    return span_df
```

```python
span_dfs = {}

for ds_key, trace_df in trace_dfs.items():
    print(f"\nParsing spans for {ds_key} ...")
    span_df = parse_trace_df_to_spans(trace_df)

    # basic cleaning
    if not span_df.empty:
        span_df["startTime"] = pd.to_numeric(span_df["startTime"], errors="coerce")
        span_df["duration"] = pd.to_numeric(span_df["duration"], errors="coerce")

        # Jaeger timestamps are often in microseconds
        # we keep raw for now, but also create readable datetime
        span_df["startTime_dt"] = pd.to_datetime(span_df["startTime"], unit="us", errors="coerce")

        span_df = span_df.drop_duplicates(subset=["traceID", "spanID"])

    span_dfs[ds_key] = span_df

    print("Shape:", span_df.shape)
    display(span_df.head(5))
```

```python

def mark_root_spans(span_df):
    df = span_df.copy()

    # Build trace -> set(spanIDs)
    trace_to_spans = df.groupby("traceID")["spanID"].apply(set).to_dict()

    def is_root(row):
        parent = row["parentSpanID"]
        trace_id = row["traceID"]

        # case 1: explicit null parent
        if pd.isna(parent) or parent in [None, "None", "<NA>"]:
            return True

        # case 2: parent not present in this trace
        if parent not in trace_to_spans.get(trace_id, set()):
            return True

        return False

    df["is_root"] = df.apply(is_root, axis=1)
    return df
```

```python
def parent_relationship_report(df, ds_name="dataset"):
    print(f"\n===== PARENT RELATIONSHIP REPORT: {ds_name} =====")

    total_spans = len(df)
    root_spans = df["is_root"].sum()
    non_root_spans = total_spans - root_spans

    print("Total spans:", total_spans)
    print("Root spans:", int(root_spans))
    print("Non-root spans:", int(non_root_spans))

    # parent missing among non-roots should ideally be 0
    problematic_non_root = df[
        (~df["is_root"]) &
        (df["parentSpanID"].isna() | (df["parentSpanID"].astype(str).isin(["None", "<NA>"])))
    ]

    print("Problematic non-root spans with null parent:", len(problematic_non_root))

for ds_key, df in span_dfs.items():
    parent_relationship_report(df, ds_key)
```

```python
def parse_tag_list(tag_list):
    """
    Convert list of tag dicts like:
    [{'key': 'http.status_code', 'type': 'int64', 'value': 200}, ...]
    into:
    {'http.status_code': 200, ...}
    """
    if not isinstance(tag_list, list):
        return {}

    out = {}
    for t in tag_list:
        if isinstance(t, dict):
            key = t.get("key")
            val = t.get("value")
            if key is not None:
                out[key] = val
    return out
```

```python
def enrich_with_tag_features(df):
    df = df.copy()

    parsed_tags = df["tags"].apply(parse_tag_list)

    df["http_status_code"] = parsed_tags.apply(lambda x: x.get("http.status_code"))
    df["error_tag"] = parsed_tags.apply(lambda x: x.get("error"))
    df["http_method"] = parsed_tags.apply(lambda x: x.get("http.method"))
    df["component"] = parsed_tags.apply(lambda x: x.get("component"))
    df["span_kind"] = parsed_tags.apply(lambda x: x.get("span.kind"))
    df["peer_service"] = parsed_tags.apply(lambda x: x.get("peer.service"))

    # useful booleans for later
    df["is_error"] = df["error_tag"].notna() | (
        pd.to_numeric(df["http_status_code"], errors="coerce") >= 400
    )

    return df

for ds_key, df in span_dfs.items():
    span_dfs[ds_key] = enrich_with_tag_features(df)
    print(f"{ds_key} enriched.")
```

```python
def trace_graph_sanity(df, ds_name="dataset"):
    print(f"\n===== TRACE GRAPH SANITY: {ds_name} =====")

    # child counts
    child_counts = df.groupby("parentSpanID").size().reset_index(name="num_children")

    print("Average children per parent span:", child_counts["num_children"].mean())
    print("Max children for a parent span:", child_counts["num_children"].max())

    roots_per_trace = df.groupby("traceID")["is_root"].sum()
    print("\n% traces with exactly 1 root:",
          round((roots_per_trace == 1).mean() * 100, 2))
    print("% traces with >1 root:",
          round((roots_per_trace > 1).mean() * 100, 2))

for ds_key, df in span_dfs.items():
    trace_graph_sanity(df, ds_key)
```

```python
from collections import defaultdict

def build_trace_lookup(span_df):
    """
    Returns:
      trace_to_spans: dict[traceID] -> dataframe of spans
      trace_parent_map: dict[traceID] -> {spanID: parentSpanID}
      trace_service_map: dict[traceID] -> {spanID: serviceName}
      trace_root_spans: dict[traceID] -> list of root spanIDs
    """
    trace_to_spans = {}
    trace_parent_map = {}
    trace_service_map = {}
    trace_root_spans = {}

    for trace_id, g in span_df.groupby("traceID"):
        g = g.sort_values("startTime").copy()

        parent_map = dict(zip(g["spanID"], g["parentSpanID"]))
        service_map = dict(zip(g["spanID"], g["serviceName"]))
        roots = g.loc[g["is_root"], "spanID"].tolist()

        trace_to_spans[trace_id] = g
        trace_parent_map[trace_id] = parent_map
        trace_service_map[trace_id] = service_map
        trace_root_spans[trace_id] = roots

    return trace_to_spans, trace_parent_map, trace_service_map, trace_root_spans
```

```python

def get_service_path(trace_id, span_id, parent_map, service_map, max_hops=100):
    """
    Reconstruct path from root -> current span as tuple of service names.
    """
    path = []
    current = span_id
    visited = set()
    hops = 0

    while current is not None and current not in visited and hops < max_hops:
        visited.add(current)
        svc = service_map.get(current)
        if svc is not None and str(svc) != "<NA>":
            path.append(str(svc))

        parent = parent_map.get(current)
        if pd.isna(parent) or parent in [None, "None", "<NA>"]:
            break

        current = parent
        hops += 1

    path.reverse()
    return tuple(path)
```

```python
def extract_stv_events(span_df):
    """
    Creates a long form list of STV events:
      traceID | serviceName | service_path | duration | operationName
    """
    events = []

    trace_to_spans, trace_parent_map, trace_service_map, trace_root_spans = build_trace_lookup(span_df)

    for trace_id, g in trace_to_spans.items():
        parent_map = trace_parent_map[trace_id]
        service_map = trace_service_map[trace_id]

        for _, row in g.iterrows():
            span_id = row["spanID"]
            service_name = row["serviceName"]
            duration = row["duration"]
            operation_name = row["operationName"]

            service_path = get_service_path(trace_id, span_id, parent_map, service_map)

            events.append({
                "traceID": trace_id,
                "spanID": span_id,
                "serviceName": service_name,
                "service_path": service_path,
                "duration": duration,
                "operationName": operation_name,
            })

    return pd.DataFrame(events)
```

```python
def build_stv_vocab(stv_events_df):
    """
    Vocabulary = all unique (serviceName, service_path) pairs from train set
    """
    keys = list(zip(stv_events_df["serviceName"], stv_events_df["service_path"]))
    unique_keys = sorted(set(keys), key=lambda x: (str(x[0]), str(x[1])))
    key_to_idx = {k: i for i, k in enumerate(unique_keys)}
    return unique_keys, key_to_idx

# Re-ensure constants are defined in case of kernel restart
TRAIN_DS = "dataset_2"
TEST_DS_1 = "dataset_1"
TEST_DS_2 = "dataset_3"

train_events = stv_event_dfs[TRAIN_DS]
stv_vocab, stv_key_to_idx = build_stv_vocab(train_events)

print("STV vocabulary size:", len(stv_vocab))
print("Sample keys:")
for k in stv_vocab[:10]:
    print(k)
```

```python
import numpy as np

def events_to_stv_matrix(stv_events_df, key_to_idx, use_log_duration=True):
    trace_ids = sorted(stv_events_df["traceID"].unique())
    trace_to_row = {tid: i for i, tid in enumerate(trace_ids)}

    X = np.full((len(trace_ids), len(key_to_idx)), -1.0, dtype=np.float32)

    for _, row in stv_events_df.iterrows():
        tid = row["traceID"]
        key = (row["serviceName"], row["service_path"])

        if key not in key_to_idx:
            continue

        r = trace_to_row[tid]
        c = key_to_idx[key]

        val = float(row["duration"])
        if use_log_duration:
            val = np.log1p(max(val, 0.0))

        X[r, c] = val

    return trace_ids, X
```

```python
stv_matrices = {}

for ds_key, ev in stv_event_dfs.items():
    trace_ids, X = events_to_stv_matrix(ev, stv_key_to_idx, use_log_duration=True)
    stv_matrices[ds_key] = {
        "trace_ids": trace_ids,
        "X": X
    }

    print(f"{ds_key}: X shape = {X.shape}")
```
